### Example workflows
- One ponded depth field

```python
import geopandas
import rasterio

from depth_downscaler import downscale_vol_elev
```

```python
elev_path = "data/USGS_one_meter_hand.tif"
mesh_path = "data/ats_village_ponded_clip.gpkg"
volume_geometry_path = "data/segment_catchment_village.gpkg"
```

```python
inundated, inun_profile = downscale_vol_elev(
    elev_path, volume_geometry_path, mesh_path, "ponded_wat"
)

with rasterio.open(f"inun.tif", "w", **inun_profile) as ds:
    ds.write(inundated, 1)
    print(f"wrote inun.tif")
```

```python
mesh_path = "data/pd_village_slices/slices_filt.shp"

columns = geopandas.read_file(mesh_path, rows=1).columns
ponded_water_felds = columns[1:-1].tolist()
print(ponded_water_felds)
```

```python
for pwf in ponded_water_felds:
    inundated, inun_profile = downscale_vol_elev(
        elev_path, volume_geometry_path, mesh_path, pwf
    )
    with rasterio.open(f"inun_{pwf}.tif", "w", **inun_profile) as ds:
        ds.write(inundated, 1)
        print(f"wrote inun_{pwf}.tif\n")
```
